```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
popus = pd.read_csv("/content/us_pop_by_state.csv")
popus.head()
```

```python
newpop = popus.iloc[1:20]
newpop.head()
```

```python
plt.figure(figsize=(12,6))
plt.plot(newpop['percent_of_total'],newpop['state_code'])
plt.xlabel('percent_of_total')
plt.ylabel('state_code')
plt.title('Date over country')
plt.grid(True)
plt.show()
```

these are the 20 most populated state in USA and while they are the most populated there's not a single one with over 1% of the total population

```python
popglobal = pd.read_csv("/content/popolazione-globale-per-paese-1950-2024.csv")
popglobal.head()
```

```python
afgpop = popglobal.iloc[54:75]
afgpop.head()
```

```python
plt.figure(figsize=(12,6))
plt.plot(afgpop['year'],afgpop['population'])
plt.xlabel('year')
plt.ylabel('population')
plt.title('Afghanistan')
plt.grid(True)
plt.show()
```

Afghanistan's population has had a good trend going from about 23 million to 42 million in just 20 years almost doubling their population

```python
albpop = popglobal.iloc[130:150]
albpop.head()
```

```python
plt.figure(figsize=(12,6))
plt.plot(albpop['year'],albpop['population'])
plt.xlabel('year')
plt.ylabel('population')
plt.title('Albania')
plt.grid(True)
plt.show()
```

Albaina's population has reduced by over 1million turning a population of over 3.8 million to just about 2.7 million in 20 years
