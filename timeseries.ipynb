```python
%pip install statsmodels

import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import adfuller
```

```python
popus = pd.read_csv("/content/2024_2026_bitcoin.csv")
popus.head()
```

```python
popus.tail()
```

```python
# Test to see if stational
# if stationary less than 0 if not would be greater than 0

popus.columns = [col.lower() for col in popus.columns]
```

```python
popus.head()
```

```python
popus['close'].isna().sum()
```

```python
popus.dropna(inplace=True)
```

```python
price_test_stationary = adfuller(popus['close'])
```

```python
print(f"The value is {price_test_stationary[0]}")
print(f"the p value {price_test_stationary[1]}")
```

```python
log_returns = np.log(popus['close'] / popus['close'].shift(1))
```

```python
log_returns.isna().sum()
```

```python
log_returns.dropna(inplace=True)
```

```python
log_returns_test_stationary = adfuller(log_returns)
```

```python
print(f"The value is {log_returns_test_stationary[0]}")
print(f"the p value {log_returns_test_stationary[1]}")
```
